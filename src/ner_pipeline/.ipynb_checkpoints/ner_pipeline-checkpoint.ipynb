```python
import pandas as pd

# load mapping file
mapping_df = pd.read_csv("data/entity_stock_mapping.csv")

# convert to dictionary
entity_to_stock = dict(zip(mapping_df["entity"], mapping_df["stock"]))

print(entity_to_stock)
```

```python
def normalize_entity(text):

    text = text.lower()

    remove_words = [
        "limited",
        "ltd",
        "ltd.",
        "inc",
        "inc.",
        "corp",
        "corporation"
    ]

    for word in remove_words:
        text = text.replace(word, "")

    return text.strip()
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")

def detect_stocks(news):

    detected_stocks = set()

    news_lower = normalize_entity(news)

    for entity, stock in entity_to_stock.items():

        normalized_entity = normalize_entity(entity)

        if normalized_entity in news_lower:
            detected_stocks.add(stock)

    doc = nlp(news)

    for ent in doc.ents:

        entity = normalize_entity(ent.text)

        if entity in entity_to_stock:
            detected_stocks.add(entity_to_stock[entity])

    return list(detected_stocks)
```

```python
news = "Reliance Industries Limited plans new refinery."

stocks = detect_stocks(news)

print("Detected stocks:", stocks)
doc = nlp(news)

for ent in doc.ents:
    print(ent.text, ent.label_)
```
